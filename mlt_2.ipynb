# to implement simple linear regression we perform some steps
''''1. importing required libarires
2.read the data
3. model selection and fitting
4.predicting the value using  linear equation y=mx+c
5. visualizing the data'''


```python
#1. importing libraraies
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#2. reading the data
data={
        "hours":[1,2,3,4,5],
        "marks":[50,60,70,80,100]
}
df=pd.DataFrame(data)
print(df)
```

```python
X=df[["hours"]].values
y=df[["marks"]].values
```

Model fitting
equation for linear regression is y=mx+c what it is y/x means,change in dependent variable when there is unit change in independent variable .
what is intercept= even in the absence of x there iswill be some value i.e even if the student is not studying he/she will be getting some marks that is bascially intercept according to this example

so basicallly here we are finding change in marks when theses is unit change in no of hours studied.

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(X,y)
```

```python
slope=model.coef_[0]
intercept=model.intercept_
```

```python
#f-string (formatted string letral in python)
#.2f this is a format specifier used with in the curly braces of an f-string
print(f"slope(m):{slope}")
print(f"intercept(b):{intercept}")
print(f"equation of the regression line : y={slope[0]:.2f}x+{intercept[0]:.2f}")
```

```python
df['predicted marks']=model.predict(X)
```

```python
#HERE KEEP IN MIND THE DATA WHICH WE HAVE PROVIDED TO THE MODEL IS ALSO LINEAR
df.head()
```

```python
# visualizing
plt.scatter(df['hours'],df['marks'],color='blue',label='actual marks')
```

```python
#plotting the equation
plt.plot(df['hours'],df['predicted marks'],color='red',label='regression line')
plt.xlabel('hours studied')
plt.ylabel('marks')
plt.title('linear regression :hours vs marks')
plt.legend()
plt.show()
```
