step3: data preprocessing
data clening "removing impurities" imputation

```python
titanic_data.isnull()
```

```python
#imputation
titanic_data.drop('Cabin',axis=1,inplace=True)
```

```python
titanic_data.head(2)
```

```python
titanic_data.dropna(inplace=True)
```

```python
titanic_data.isnull().sum()
```

converting data from one form to another

```python
titanic_data.head(5)
# we will first convert categorical var into dummmy var and these useb then as logistic reg only takes two input
```

```python
#encoding
#using logistic regression we willpredict how many people survived and how many died in titanic ship. here 1 survived and 0 died
pd.get_dummies(titanic_data['Sex']).astype(int)
```

```python
S=pd.get_dummies(titanic_data['Sex'],drop_first=True).astype(int)
S.head(4)
```

```python
# conversion of embaRKED COLUMN FROM ONE FARE TO ANOTHER
#Q=QUINSTOWN,S=SOUTHAM,C=CHAMBOR
E=pd.get_dummies(titanic_data['Embarked'],drop_first=True).astype(int).astype(int)
E.head(4)
```

```python
#conversion of pclass from one form  to another
P=pd.get_dummies(titanic_data['Pclass'],drop_first=True).astype(int).astype(int)
P.head(4)
```

```python
#concatinate all the new rows in the dataset
titanic_data=pd.concat([titanic_data,S,E,P],axis=1)
```

```python
titanic_data.head(5)
```

```python
titanic_data.drop(['Sex','Embarked','PassengerId','Name','Ticket','Pclass'],axis=1,inplace=True)
```

```python
#our final dataset
titanic_data.head(5)
```

step 4: split training and testing data

```python
#train data
x=titanic_data.drop('Survived',axis=1)#accept survived all the other columnsare indepvar
y=titanic_data['Survived']
```

```python
#splitting the data
from sklearn.model_selection import train_test_split
```

```python
train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=1)
```

implementing logistic regression
!!!!! creating our model

```python
from sklearn.linear_model import LogisticRegression
```

```python
model =LogisticRegression()
```

```python
X_train.columns =X_train.columns.astype(str)
model.fit(X_train,y_train)
```

```python
X_test.columns =X_test.columns.astype(str)
predictions = model.predict(X_test)
```

STEP 5: PREDICTIONS

```python
print("predictions:")
print(predictions)
```

```python
import matplotlib.pyplot as plt
#assuming u already have your trained model and test data
#model.fit(X_train,y_train)
predictions =model.predict(X_test)
#plot predicted classes vs the first feature of X_tset
plt.plot(X_test.iloc[:,0],predictions,marker='o',linestyle='-',color='blue')
plt.xlabel('featured')
plt.ylabel('survived')
plt.title('logistic regression predictions')
plt.grid(True)
plt.show()
```

```python
from sklearn.metrics import confusion_matrix
```

```python
print(confusion_matrix(y_test,predictions))
```

step 6:checking the accuracy

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test,predictions)
```
